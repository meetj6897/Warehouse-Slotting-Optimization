```python
def custom_sort(coordinates):
    # Sort based on the first coordinate
    coordinates.sort(key=lambda x: x[0])
    # print('c',coordinates)

    uniq=set()
    for cord in coordinates:
        uniq.add(cord[0])
    uniq=list(uniq)
    # print(uniq)
    # Group coordinates with the same first coordinate
    groups = []
    current_group = []
    prev_x = None

    for coord in coordinates:
        if coord[0] != prev_x:
            if current_group:
                groups.append(current_group)
            current_group = []
            prev_x = coord[0]
        current_group.append(coord)

    if current_group:
        groups.append(current_group)
    # print('g',groups)

    # Sort within each group
    for group in groups:
        for i, coord in enumerate(group):
            # print('kkkkkkk',i, coord )
            if uniq.index(coord[0]) % 2 == 0:
                group.sort(key=lambda x: x[1])
            else:
                group.sort(key=lambda x: x[1], reverse=True)

    # Flatten the sorted groups
    sorted_coordinates = [coord for group in groups for coord in group]

    return sorted_coordinates



def s_dist(order_list):

  uniq=set()
  for cord in order_list:
      uniq.add(cord[0])
  uniq=list(uniq)
  # print(uniq)
  if len(uniq)%2==0: # means if the even aisle visiting then totla verical mement is equal to traveral of all aisle so just multiply the verticle length
    total_verical_movement=len(uniq)*25  # assuming the vericle length is 25

    #horizontal_movement=

    dist_origin_to_first=order_list[0][0]+order_list[0][1]
    dist_last_to_origin=order_list[len(order_list)-1][0]  # +order_list[len(order_list)-1][1] removing this bcause already covered in vertical distance
    total_dist_horizontal=0
    for i in range(len(order_list)-1):
      a=order_list[i] #(1,3)
      b=order_list[i+1]  #(2,4)
      horizontal_dist=b[0]-a[0] # hosrizontal distance

      total_dist_horizontal= total_dist_horizontal + horizontal_dist

    total_traversal_distance_sshape= dist_origin_to_first + total_verical_movement + total_dist_horizontal + dist_last_to_origin

    return total_traversal_distance_sshape

  else:
    """"# when the total number of visiting aisle is odd then we traverse all aisle except last one
    in the last one we only traverse till the order position not traverse entire  aisle """
    total_verical_movement=(len(uniq)-1)*25 + order_list[len(order_list)-1][1] # assuming the vericle length is 25

    #horizontal_movement=

    dist_origin_to_first=order_list[0][0]+order_list[0][1]
    dist_last_to_origin=order_list[len(order_list)-1][0]+order_list[len(order_list)-1][1] # here we retain + order_list[len(order_list)-1][1] term because it traverse two time on same path for returning
    total_dist_horizontal=0
    for i in range(len(order_list)-1):
      a=order_list[i] #(1,3)
      b=order_list[i+1]  #(2,4)
      horizontal_dist=b[0]-a[0] # hosrizontal distance

      total_dist_horizontal= total_dist_horizontal + horizontal_dist
    total_traversal_distance_sshape= dist_origin_to_first + total_verical_movement + total_dist_horizontal + dist_last_to_origin

    return total_traversal_distance_sshape
```

```python
def create_child(parent1, parent2):
    n = len(parent1)
    start = np.random.randint(0, n)
    end = np.random.randint(start + 1, n + 1)
    # print(start,end)
    child = [-1] * n
    for i in range(start, end):
        child[i] = parent1[i]
    remaining = [city for city in parent2 if city not in child]
    j = 0
    for i in range(n):
        if child[i] == -1:
            child[i] = remaining[j]
            j += 1
    return child
```

```python
# Mutation (Swap Mutation)
def mutation(individual, mutation_rate):
    if np.random.rand() < mutation_rate:
        idx1, idx2 = np.random.choice(len(individual), 2, replace=False)
        individual[idx1], individual[idx2] = individual[idx2], individual[idx1]
    return individual
```
